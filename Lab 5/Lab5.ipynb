```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

### Task 1

```python
img = cv2.imread('cat.jpg')
assert img is not None, "file could not be read, check with os.path.exists()"

img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
kernel = np.ones((7,7),np.float32)/49
dst = cv2.filter2D(img,-1,kernel)
 
plt.subplot(121),plt.imshow(img),plt.title('Original')
plt.xticks([]), plt.yticks([])
plt.subplot(122),plt.imshow(dst),plt.title('Averaging')
plt.xticks([]), plt.yticks([])
plt.show()
```

### Task 2

```python
blur = cv2.GaussianBlur(img,(5,5),0)
blur_big = cv2.GaussianBlur(img, (21,21),0)

plt.subplot(131),plt.imshow(img),plt.title('Original')
plt.xticks([]), plt.yticks([])
plt.subplot(132),plt.imshow(blur),plt.title('5x5')
plt.xticks([]), plt.yticks([])
plt.subplot(133),plt.imshow(blur_big),plt.title('21x21')
plt.xticks([]), plt.yticks([])
plt.show()
```

### Task 3

```python
img_float32 = img.astype(np.float32)/ 255.0
lapl_img = cv2.Laplacian(img_float32,cv2.CV_32F)
sharpened = np.clip(img_float32 - lapl_img, 0, 1)
plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.title("Original")
plt.imshow(img_float32)
plt.axis("off")

plt.subplot(1,3,2)
plt.title("Laplacian")
plt.imshow(lapl_img)
plt.axis("off")

plt.subplot(1,3,3)
plt.title("Sharpened")
plt.imshow(sharpened)
plt.axis("off")

plt.show()
```
