````python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.cm as cm
from skimage import data
````

### Task#1: Load/read images cat.jpeg and cameraman.tif [1 mark]

````python
# Load/read images cat.jpeg and cameraman.tif [1 mark]
# Using scikit-image sample images: chelsea (cat), camera (cameraman / cameraman.tif).
cat_rgb = data.chelsea()
cam = data.camera()
cat = cv2.cvtColor(cat_rgb, cv2.COLOR_RGB2BGR)

print("cat shape:", cat.shape, "| dtype:", cat.dtype)
print("cameraman shape:", cam.shape, "| dtype:", cam.dtype)
````

### Task#2: Add the two images, and display the result [2 marks]

````python
# Add the two images, and display the result [2 marks]
# Resize cat to match cameraman; use grayscale so addition is defined on one channel.
if cam.ndim == 3:
    h, w = cam.shape[:2]
    cam_gray = cv2.cvtColor(cam, cv2.COLOR_BGR2GRAY)
else:
    h, w = cam.shape
    cam_gray = cam

cat_resized = cv2.resize(cat, (w, h), interpolation=cv2.INTER_AREA)
cat_gray = cv2.cvtColor(cat_resized, cv2.COLOR_BGR2GRAY)

summed = np.clip(
    cat_gray.astype(np.float32) + cam_gray.astype(np.float32),
    0,
    255,
).astype(np.uint8)

plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(cat_resized, cv2.COLOR_BGR2RGB))
plt.title("Cat (resized)")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(cam_gray, cmap="gray")
plt.title("Cameraman")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(summed, cmap="gray")
plt.title("Sum (clipped to [0, 255])")
plt.axis("off")

plt.tight_layout()
plt.show()
````

### Task#3: Using kernel1 and kernel2 apply convolution to cameraman.tif, and display the output image [2 marks]

```text
kernel1 = [[-1, -2, -1], [0, 0, 0], [1, 2, 1]]
kernel2 = [[-1, 0, 1], [-2, 0, 2], [-1, 0, 1]]
```

````python
# Using kernel1 and kernel2 apply convolution to cameraman.tif, and display the output image [2 marks]
# Kernels fixed as given in the quiz (do not edit coefficients).
kernel1 = [[-1, -2, -1],
           [0, 0, 0],
           [1, 2, 1]]

kernel2 = [[-1, 0, 1],
           [-2, 0, 2],
           [-1, 0, 1]]

if cam.ndim == 3:
    cam_gray = cv2.cvtColor(cam, cv2.COLOR_BGR2GRAY)
else:
    cam_gray = cam.copy()

cam_f = cam_gray.astype(np.float32)
k1 = np.asarray(kernel1, dtype=np.float32)
k2 = np.asarray(kernel2, dtype=np.float32)
out1 = cv2.filter2D(cam_f, -1, k1)
out2 = cv2.filter2D(cam_f, -1, k2)


def normalize_for_display(img):
    """Scale signed filter output to [0, 255] for visualization."""
    i = img.astype(np.float32)
    i_min, i_max = i.min(), i.max()
    if i_max > i_min:
        i = (i - i_min) * (255.0 / (i_max - i_min))
    else:
        i = np.zeros_like(i)
    return i.astype(np.uint8)


plt.figure(figsize=(14, 4))
plt.subplot(1, 3, 1)
plt.imshow(cam_gray, cmap="gray")
plt.title("Cameraman (input)")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(normalize_for_display(out1), cmap="gray")
plt.title("Response to kernel1")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(normalize_for_display(out2), cmap="gray")
plt.title("Response to kernel2")
plt.axis("off")

plt.tight_layout()
plt.show()
````
